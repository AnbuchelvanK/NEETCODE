# Top K Frequent Elements - 19 Jan

```python
class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:

        dict1 = {}
        lis = []
        for i in nums:
            if i not in dict1:
                dict1[i] = 1

            else:
                dict1[i] += 1


        sorted_items = list(sorted(dict1.items(), key=lambda x: x[1], reverse=True))
        for i in range(k):
            lis.append(sorted_items[i][0])
        return lis
```

Time Complexity : O(n log n)

Space Complexity :O(n)
