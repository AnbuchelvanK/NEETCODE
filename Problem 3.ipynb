# PROBLEM 3 - 2nd Jan - Two Sum

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:

        d = {}
        i = 0

        while i < len(nums):
            n = nums[i]
            if (target-n) in d:
                return [d[target-n], i]

            d[n] = i
            i += 1
```

Time Complexity : O(n)

Space Complexity :O(n)
