# PROBLEM 1 - 17th Dec - Contains Duplicate

```python
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        dic = {}
        for i in nums:
            if i not in dic:
                dic[i] = 1
            else:
                return True
        return False
```

Time Complexity : O(n)

Space Complexity :O(n)
